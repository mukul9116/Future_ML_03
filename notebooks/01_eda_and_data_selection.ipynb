```python
for title in ['Quality Engineer Job in Durham', 'Packaging Engineer Job in Chicago', 'Electrical Engineer']:
  text = df_jobs[df_jobs['job_title'] == title]['job_description'].iloc[0]
  print(f"{title} ----> ")
  print(text[:600])
  print()
```

```python
selected_titles = {
    'INFORMATION-TECHNOLOGY':'IT Support Technician Job in Madison',
    'FINANCE':'Senior Accountant/Analyst Job in Denver',
    'ENGINEERING':'Quality Engineer Job in Durham',
    'SALES':'Sales Professional Job in Las Vegas',
    'HR':'Human Resources Manager Job in Dallas',
    'HEALTHCARE':'Registered Nurse - Clinic Job in Houston'

}
```

```python
for category, title in selected_titles.items():
  matches = df_jobs[df_jobs['job_title']==title]
  print(category,' :',title, ' --> matches found', matches.shape[0])
```

```python
jd_rows =[]
for category, title in selected_titles.items():
  rows = df_jobs[df_jobs['job_title']==title].iloc[0]
  jd_rows.append({'category': category, 'job_title':rows['job_title'], 'job_description': rows['job_description']})

df_jd = pd.DataFrame(jd_rows)
print(df_jd.shape)
df_jd.to_pickle('/content/drive/MyDrive/Colab Notebooks/Future_ML_03/data/selected_jobs_description.pkl')
print('done')
```
