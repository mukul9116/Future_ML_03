Validation through Correlation

```python
df_cleaned_resume['resume_length'] = df_cleaned_resume['cleaned_resume'].apply(lambda x: len(x.split()))

correlation = np.corrcoef(df_cleaned_resume['resume_length'], df_cleaned_resume['best_match_score'])[0,1]
print(f"corelation between resume length and similarity score : {correlation:.3f}")
```

```python
df_cleaned_resume['num_extracted_skills'] = df_cleaned_resume['extracted_skills'].apply(len)

correlation = np.corrcoef(df_cleaned_resume['resume_length'], df_cleaned_resume['num_extracted_skills'])[0,1]
print(f"Correlation between resume length and number of skills extracted: {correlation:.3f}")
```
